```python
import numpy as np
import pandas as pd

df = pd.read_csv(r'C:\Users\Marcus\Desktop\python pandas\superstore_data\version_control\superstore_clean.csv')

df['Order Date'] = pd.to_datetime(df['Order Date'])

df
```

```python
df.info()
```

```python
df.groupby(
    'Ship Mode'
)[[
    'Sales'
]].agg(
    Total_Sales = ('Sales','sum'),
    Average_Sales = ('Sales','mean'),
    Median_Sales = ('Sales','median'),
    Minimum_Sales = ('Sales','min'),
    Maximum_Sales = ('Sales','max'),
    Number_of_Transactions = ('Sales','count'),
).round(2)
```

```python
df.groupby(
    'Ship Mode'
)[[
    'Sales',
    'Quantity',
    'Discount',
    'Profit'
]].agg(
    Total_Units_Sold = ('Quantity','sum'),
    Average_Units_Sold = ('Quantity','mean'),
    Median_Quantity = ('Quantity','median'),
    Minimum_Quantity = ('Quantity','min'),
    Maximum_Quantity = ('Quantity','max'),
).round(2)
```

```python
df.groupby(
    'Ship Mode'
)[[
    'Discount'
]].agg(
    Total_Discount_Pct = ('Discount','sum'),
    Average_Discount = ('Discount','mean'),
    Median_Discount = ('Discount','median'),
    Minimum_Discount = ('Discount','min'),
    Maximum_Discount = ('Discount','max')
).round(2)
```

```python
df.groupby(
    'Ship Mode'
)[[
    'Profit'
]].agg(
    Total_Profit = ('Profit','sum'),
    Average_Profit = ('Profit','mean'),
    Median_Profit = ('Profit','median'),
    Minimum_Profit = ('Profit','min'),
    Maximum_Profit = ('Profit','max'),
    Standard_Deviation = ('Profit','std')
).round(2)
```
