```python
df['Customer Name']=df['Customer Name'].str.lstrip().str.rstrip()
```

```python
df['Customer Name'] = df['Customer Name'].fillna('Unknown')
```

```python
df['Customer Name'] = df['Customer Name'].replace('Nan','Unknown')
```
