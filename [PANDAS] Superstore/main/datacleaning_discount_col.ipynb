```python
df['Discount'] = df['Discount'].replace(
    r'[-|?|Unknown]','0.00',
    regex=True
)
```

```python
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce').round(2)
```

```python
df['Discount'].replace(
    {
        20.00:0.20,
        10.00:0.10,
        40.00:00.40,
        70.00:00.70,
        60.00:00.60,
        50.00:00.50,
        17:
    }
)
```

```python
condition = df['Discount'] > 1.00

df.loc[condition, 'Discount'] = df.loc[condition, 'Discount'] / 100
```
