```python
import numpy as np
import pandas as pd

df = pd.read_csv(r'C:\Users\Marcus\Desktop\python pandas\superstore_data\version_control\superstore_clean.csv')

df
```

```python
df.info()
```

```python
df.groupby(
    'Order Priority'
)[[
    'Sales',
    'Quantity',
    'Discount',
    'Profit'
]].agg(
    Total_Sales = ('Sales','sum'),
    Average_Sales = ('Sales','mean'),
    Minimum_Sales = ('Sales','min'),
    Maximum_Sales = ('Sales','max'),
	Sales_Standard_Deviation = ('Sales','std'),
    Number_of_Transaction = ('Sales','count'),

    Total_Units_Sold = ('Quantity','sum'),
    Average_Units_Sold = ('Quantity','mean'),
    Minimum_Quantity = ('Quantity','min'),
    Maximum_Quantity = ('Quantity','max'),
    Quantity_Standard_Deviation = ('Quantity','std'),

    Total_Discount = ('Discount','sum'),
    Average_Discount = ('Discount','mean'),
    Minimum_Discount = ('Discount','min'),
    Maximum_Discount = ('Discount','max'),
    Discount_Standard_Deviation = ('Discount','std'),

    Total_Profit = ('Profit','sum'),
    Average_Profit = ('Profit','mean'),
    Minimum_Profit = ('Profit','min'),
    Maximum_Profit = ('Profit','max'),
    Profit_Standard_Deviation = ('Profit','std')
).round(2)
```
